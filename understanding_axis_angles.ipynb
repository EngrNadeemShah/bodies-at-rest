# 🌀 Understanding Axis-Angle Representation (SMPL Context)

## 📘 What is Axis-Angle?
- **Axis-Angle** is a way to represent **3D rotations**.
- It encodes rotation using **3 values**:
  - A **unit vector** → the axis of rotation
  - A **scalar** (in radians) → the angle of rotation around that axis
- These are combined as a **3D vector**:  
  `rotation_vector = axis * angle` -> (also known as a **Rodrigues vector**)

## 🧠 Axis-Angle in SMPL Model

- SMPL uses axis-angle to represent joint rotations.
- There are **24 joints**, each with **3 values** → total **72D pose vector**:
	``` python
	pose = [joint0_x, joint0_y, joint0_z, ..., joint23_z]  # 72 values
	```
- This 72D vector is passed through the Rodrigues formula (from OpenCV, PyTorch3D, or similar) to get rotation matrices for each joint.

## 🧪 Minimal Python Example: Axis-Angle ➡️ Rotation Matrix

````python
import numpy as np
from scipy.spatial.transform import Rotation as R
np.set_printoptions(suppress=True)

# Rotate 90 degrees (π/2) around the Y-axis
axis_angle = np.array([0, np.pi / 2, 0])

# Convert to rotation matrix
rot = R.from_rotvec(axis_angle)
rot_matrix = rot.as_matrix()

print("Axis-Angle Vector:", axis_angle)
print("Rotation Matrix:\n", rot_matrix)
````

## 🔁 Rotate a Point Using the Rotation Matrix

````python
point = np.array([1, 0, 0])  # point on X-axis
rotated_point = rot.apply(point)
print("Rotated Point:", rotated_point)
````

Expected result:

The point [1, 0, 0] rotates to [0, 0, -1] → a 90° rotation around the Y-axis

## ✅ Summary Table

| Term                | Description                                                            |
| ------------------- | ---------------------------------------------------------------------- |
| **Axis**            | Unit vector defining direction of rotation                             |
| **Angle**           | Scalar in radians for how much to rotate                               |
| **Axis-Angle (3D)** | Combined as a single 3D vector: `axis × angle`                         |
| **SMPL Pose**       | 24 joints × 3 values = **72D vector**, each 3D vector is an axis-angle |


## Rotation Vector | Rotation Matrix

````python
from scipy.spatial.transform import Rotation as R
import numpy as np

# Step 1: Create a rotation vector (axis × angle)
rotvec = np.array([0, np.pi/2, 0])  # 90° around Y-axis

# Step 2: Apply Rodrigues formula to get rotation matrix
rot = R.from_rotvec(rotvec)
rot_matrix = rot.as_matrix()

# Step 3: Inverse: go back to rotation vector
back_to_rotvec = R.from_matrix(rot_matrix).as_rotvec()

print("Rotation Vector:", rotvec)
print("Rotation Matrix:\n", rot_matrix)
print("Back to Rotation Vector:", back_to_rotvec)
````

````python
print("Rotation Matrix:\n", rot.as_matrix())
print("Quaternion:", rot.as_quat())
# print("Euler Angles (xyz, radians):", rot.as_euler('xyz'))
print("Axis-Angle (Rodrigues vector):", rot.as_rotvec())
````
