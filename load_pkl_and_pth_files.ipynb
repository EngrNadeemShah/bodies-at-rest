```python
import pickle

# Replace 'your_file.pkl' with the path to your .pkl file
pkl_file_path = 'runs/run_20250525_190443/checkpoint_losses__epoch_6__valid_loss_1655.0625.pkl'
with open(pkl_file_path, 'rb') as f:
    data = pickle.load(f)

print(f"data keys: {data.keys()}")

print("losses:")
for key, value in data.items():
    if isinstance(value, list):
        print(f"{key}: {len(value)} items")
    else:
        print(f"{key}: {value}")
print("\nExample of losses:")
for key, value in data.items():
    if isinstance(value, list) and len(value) > 0:
        print(f"{key}: {value[:5]}")  # Print first 5 items of each list
    elif not isinstance(value, list):
        print(f"{key}: {value}")
```
