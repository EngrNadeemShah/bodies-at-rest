### To read a specific dataset

### Shapes of Inputs and Outputs

```python
# Open the HDF5 file in read mode
with h5py.File(hdf5_file_path, 'r') as f:
    def print_hdf5_structure(g, prefix=''):
        for key in g:
            item = g[key]
            if isinstance(item, h5py.Group):
                print(f"{prefix}{key}/")
                print_hdf5_structure(item, prefix + '  ')
            else:
                print(f"{prefix}{key} - shape: {item.shape}, dtype: {item.dtype}")

    print_hdf5_structure(f)
```

### Lengths of Inputs

```python
with h5py.File(hdf5_file_path, 'r') as file:

	def print_hdf5_structure(g):
		for key in g:
			item = g[key]

			if isinstance(item, h5py.Group):
				print(f"{key}", end="/")
				print_hdf5_structure(item)
			else:
				if key == 'inputs':
					print(f"\t{len(item)}")

	print_hdf5_structure(file)
```
